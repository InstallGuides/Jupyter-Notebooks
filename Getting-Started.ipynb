# Getting started with Jupyter

A notebook mixes notes like these with Python code you can run. Click the code cell below, then press Shift-Return to run it. It draws a chart with NumPy and Matplotlib, which this folder already has.


```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 200)
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.title("Your first chart")
plt.legend()
plt.show()
```

To make a notebook of your own, click **+** at the top left, then click **Jupyter-Notebooks** under **Notebook** in the Launcher. To add a package, such as pandas, open a terminal in this folder and type `uv add pandas`.
